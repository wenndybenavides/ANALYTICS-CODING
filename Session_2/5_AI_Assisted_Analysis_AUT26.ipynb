## A simple rule

> **AI can generate a suggestion. You still need to validate it.**

When AI produces code or an interpretation, ask:

1. Does the code run?
2. Is it using the correct variables?
3. Is the data being transformed appropriately?
4. Does the output actually support the claim?
5. Is the conclusion descriptive, inferential or causal?
6. What important limitations have been ignored?

### Solution

There are several issues.

**1. `bikes.dropna()` removes rows with a missing value in _any_ column.**

That is much more aggressive than necessary. If we only need `mean_temp` and `bikes_hired`, missing values in unrelated variables should not automatically remove an observation.

**2. The chart title makes a causal claim.**

A scatterplot or correlation can show an **association** between temperature and bike hires. It does not establish that higher temperature causes more hires.

**3. Other variables may be related to both temperature and bike demand.**

For example, season, year, precipitation, weekday/weekend status or longer-term changes in bike use may matter.

**4. A correlation coefficient alone does not describe the whole relationship.**

We should also inspect the scatterplot for non-linearity, changing spread, outliers and clusters.

A more defensible version would only remove missing values from the variables needed for this analysis.

A suitable interpretation would be framed as:

> The data can be used to assess whether bike hires and temperature are **associated**. The plot and correlation should be interpreted as descriptive evidence of that relationship, not as proof that temperature causes changes in bike use.

The exact strength and shape of the association should be taken from the output you observe rather than invented in advance.

# 6. A practical AI audit checklist

When using AI for analytical work, separate the audit into three areas.

### Code

* Does it run?
* Are the variable names correct?
* Is the transformation doing what the explanation says?
* Are missing values handled appropriately?
* Is unnecessary data being removed?
* Is the output reproducible?

### Analysis

* Does the method answer the actual question?
* Is the unit of analysis correct?
* Are relevant comparisons or confounding variables being ignored?
* Are summaries being calculated at the correct level?

### Interpretation

* Does the conclusion match the output?
* Is association being presented as causation?
* Is the AI inventing explanations not contained in the data?
* Are limitations acknowledged?
* Would you be comfortable defending the conclusion yourself?
